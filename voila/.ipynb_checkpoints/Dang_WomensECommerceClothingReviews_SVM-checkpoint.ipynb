# Cơ sở lý thuyết
## Giới thiệu chung
Support Vector Machine (SVM) là một thuật toán học máy có giám sát dùng để phân loại và phân tích hồi quy [1]. Một mô hình SVM là một cách biểu diễn các điểm trong không gian và lựa chọn ranh giới giữa hai thể loại sao cho khoảng cách từ các ví dụ luyện tập tới ranh giới là xa nhất có thể [2].

## Phương pháp

Để ứng dụng SVM giải quyết bài toán dự đoán phân loại đánh giá thời trang nữ, cần phân tích lựa chọn các đặc trưng liên quan đến bài toán đặt ra, đưa dữ liệu qua tiền xử lí và cuối cùng đưa dữ liệu vào thuật toán. 

### Lựa chọn các đặc trưng liên quan
Ánh xạ dạng dữ liệu ngôn ngữ sang dạng dữ liệu số. Class CountVectorizer của bộ thư viện sklearn bao hảm cả hai phần gán mỗi từ một số nguyên đại diện và đếm số lần từ xuất hiện.

```python
import pandas as pd
import numpy as np
import datetime as dt
import matplotlib.pyplot as plt
from wordcloud import WordCloud
import seaborn as sns
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix
import sklearn.metrics as mt
from plotly import tools
import plotly.offline as py
py.init_notebook_mode(connected=True)
import plotly.graph_objs as go
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')
 
df1 = pd.read_csv('Womens Clothing E-Commerce Reviews.csv')
df = df1[['Review Text','Recommended IND','Class Name','Age']]
#df1.info()
#df1.describe()
df1.head()
```

```python
# fill NA values by space
df['Review Text'] = df['Review Text'].fillna('')

# CountVectorizer() converts a collection 
# of text documents to a matrix of token counts
vectorizer = CountVectorizer()
# assign a shorter name for the analyze
# which tokenizes the string
analyzer = vectorizer.build_analyzer()
```

```python
# Rating of 4 or higher -> positive, while the ones with 
# Rating of 2 or lower -> negative 
# Rating of 3 -> neutral
#df = df[df['Rating'] != 3]
#df = df[df['Recommended IND'] == 1]
df['Sentiment'] = df['Recommended IND'] == 1
df.head()
```

```python
# split data
train_data,test_data = train_test_split(df,train_size=0.8,random_state=0)
# select the columns and 
# prepare data for the models 
X_train = vectorizer.fit_transform(train_data['Review Text'])
y_train = train_data['Sentiment']
X_test = vectorizer.transform(test_data['Review Text'])
y_test = test_data['Sentiment']
```

```python
start=dt.datetime.now()
svm = SVC()
svm.fit(X_train,y_train)
print('Elapsed time: ',str(dt.datetime.now()-start))
```

```python
print("\n Support Vector Machine (SVM)")
print(mt.classification_report(y_test, svm.predict(X_test)))
```
